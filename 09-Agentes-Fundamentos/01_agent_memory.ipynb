# PARTE 1: VARIABLES

## ¿Qué es una variable?
Una **variable** es un contenedor con nombre que almacena un valor en la memoria del computador.

## Sintaxis
```python

```

## Reglas para nombres de variables
- ✅ Pueden contener letras, números y guiones bajos `_`
- ✅ Deben empezar con letra o guión bajo
- ❌ NO pueden empezar con número
- ❌ NO pueden contener espacios
- ❌ NO pueden ser palabras reservadas (if, for, class, etc.)

## Convención de nombres (PEP 8)
- **snake_case** para variables y funciones: `user_message`, `agent_name`
- **PascalCase** para clases: `AgentMemory`, `UserProfile`
- **MAYÚSCULAS** para constantes: `MAX_MEMORY_SIZE`, `API_KEY`

# PARTE 2: TIPOS DE DATOS

## ¿Qué es un tipo de dato?
El **tipo de dato** define qué clase de valor almacena una variable y qué operaciones se pueden hacer con ella.

|--------|-------------|---------|

# PARTE 3: LISTAS (list)

## ¿Qué es una lista?
Una **lista** es una colección **ordenada** y **mutable** (modificable) de elementos.

## ¿Por qué los agentes necesitan listas?
- **Historial de mensajes**: Secuencia ordenada de la conversación
- **Lista de herramientas**: Herramientas disponibles para el agente
- **Contexto**: Últimos N mensajes relevantes

## Sintaxis
```python
# Lista vacía
lista_vacia = []

# Lista con elementos
numeros = [1, 2, 3, 4, 5]
mensajes = ["hola", "cómo estás", "bien"]

# Lista mixta (posible pero no recomendado)
mixta = [1, "texto", True, 3.14]
```

## Indexación (Acceso a elementos)
```python
lista = ["a", "b", "c", "d", "e"]
#        0    1    2    3    4    → índices positivos
#       -5   -4   -3   -2   -1    → índices negativos

lista[0]   # → "a" (primer elemento)
lista[-1]  # → "e" (último elemento)
lista[1:3] # → ["b", "c"] (slice: desde índice 1 hasta 3, sin incluir 3)
```

## Métodos principales de listas
| Método | Descripción | Ejemplo |
|--------|-------------|---------|
| `.append(x)` | Agrega x al final | `lista.append("nuevo")` |
| `.insert(i, x)` | Inserta x en posición i | `lista.insert(0, "primero")` |
| `.pop()` | Elimina y devuelve el último | `ultimo = lista.pop()` |
| `.pop(i)` | Elimina y devuelve el elemento i | `elemento = lista.pop(0)` |
| `.remove(x)` | Elimina primera ocurrencia de x | `lista.remove("valor")` |
| `.clear()` | Elimina todos los elementos | `lista.clear()` |
| `len(lista)` | Número de elementos | `len([1,2,3])` → `3` |

````python
# LISTAS COMO MEMORIA DE CONVERSACIÓN

# Inicializar historial vacío
conversation_history = []

# Agregar mensajes a la memoria (simula una conversación)
conversation_history.append("Usuario: Hola")
conversation_history.append("Agente: ¡Hola! ¿Cómo puedo ayudarte?")
conversation_history.append("Usuario: ¿Qué hora es?")
conversation_history.append("Agente: Son las 10:30 AM")
conversation_history.append("Usuario: Gracias")

print("=== HISTORIAL COMPLETO ===")
print(f"Total de mensajes: {len(conversation_history)}")
for mensaje in conversation_history:
    print(f"  • {mensaje}")

# ACCESO POR ÍNDICE
print(f"\n=== ACCESO POR ÍNDICE ===")
print(f"Primer mensaje: {conversation_history[0]}")
print(f"Último mensaje: {conversation_history[-1]}")

# SLICING - Obtener los últimos 3 mensajes (contexto reciente)
print(f"\n=== ÚLTIMOS 3 MENSAJES (CONTEXTO) ===")
recent_context = conversation_history[-3:]
for mensaje in recent_context:
    print(f"  • {mensaje}")

# VERIFICAR SI LA MEMORIA ESTÁ LLENA
MAX_MEMORY = 10
current_size = len(conversation_history)
is_memory_full = current_size >= MAX_MEMORY
print(f"\n=== ESTADO DE MEMORIA ===")
print(f"Tamaño actual: {current_size}/{MAX_MEMORY}")
print(f"¿Memoria llena?: {is_memory_full}")
````

# PARTE 4: DICCIONARIOS (dict)

## ¿Qué es un diccionario?
Un **diccionario** es una colección de pares **clave-valor**. Permite acceder a datos por nombre en lugar de por posición.

## ¿Por qué los agentes necesitan diccionarios?
- **Estado estructurado**: Guardar múltiples propiedades del agente
- **Mensajes con metadatos**: No solo el texto, también quién lo dijo, cuándo, etc.
- **Configuración**: Parámetros del agente organizados

## Sintaxis
```python
# Diccionario vacío
vacio = {}

# Diccionario con datos
persona = {
    "nombre": "María",
    "edad": 30,
    "activo": True
}
```

## Acceso a valores
```python
# Por clave
persona["nombre"]        # → "María"

# Con .get() (no lanza error si no existe)
persona.get("nombre")    # → "María"
persona.get("altura")    # → None
persona.get("altura", 0) # → 0 (valor por defecto)
```

## Métodos principales de diccionarios
| Método | Descripción | Ejemplo |
|--------|-------------|---------|
| `d["key"]` | Obtiene valor (error si no existe) | `d["nombre"]` |
| `d.get("key")` | Obtiene valor (None si no existe) | `d.get("nombre")` |
| `d["key"] = val` | Asigna o actualiza valor | `d["edad"] = 31` |
| `d.keys()` | Lista de claves | `list(d.keys())` |
| `d.values()` | Lista de valores | `list(d.values())` |
| `d.items()` | Lista de (clave, valor) | `list(d.items())` |
| `"key" in d` | ¿Existe la clave? | `"nombre" in d` → `True` |

````python
# DICCIONARIOS PARA ESTADO DEL AGENTE

# Estado completo del agente
agent_state = {
    "name": "MemoryBot",
    "version": "1.0",
    "is_active": True,
    "current_user": None,
    "interaction_count": 0,
    "last_response": None
}

print("=== ESTADO INICIAL DEL AGENTE ===")
for key, value in agent_state.items():
    print(f"  {key}: {value}")

# ACTUALIZAR estado cuando llega un usuario
agent_state["current_user"] = "user_12345"
agent_state["interaction_count"] = 1
agent_state["last_response"] = "¡Hola! ¿Cómo puedo ayudarte?"

print("\n=== ESTADO DESPUÉS DE INTERACCIÓN ===")
for key, value in agent_state.items():
    print(f"  {key}: {value}")

# ACCESO SEGURO con .get()
# Útil cuando no sabes si la clave existe
error_message = agent_state.get("last_error", "Sin errores")
print(f"\nÚltimo error: {error_message}")

# VERIFICAR si existe una clave
if "current_user" in agent_state:
    print(f"Usuario actual: {agent_state['current_user']}")
````

````python
# COMBINANDO LISTAS Y DICCIONARIOS: El patrón del State

# Este patrón es EXACTAMENTE lo que usa LangGraph:
# - Una lista de mensajes
# - Cada mensaje es un diccionario con estructura definida

# PATRÓN REAL: Lista de mensajes (como en LangGraph)
messages = []

# Agregar mensajes con estructura estándar
messages.append({
    "role": "user",           # "user", "assistant", o "system"
    "content": "Hola, soy María",
    "timestamp": "2024-01-15 10:30:00"
})

messages.append({
    "role": "assistant", 
    "content": "¡Hola María! ¿En qué puedo ayudarte?",
    "timestamp": "2024-01-15 10:30:02"
})

messages.append({
    "role": "user",
    "content": "¿Puedes recordar mi nombre?",
    "timestamp": "2024-01-15 10:30:30"
})

# PATRÓN REAL: State completo (como en LangGraph)
agent_state = {
    "messages": messages,           # Historial de conversación
    "user_id": "user_123",          # Identificador
    "current_step": "processing",   # Estado actual del flujo
    "metadata": {                   # Datos adicionales
        "session_start": "2024-01-15 10:30:00",
        "tool_calls": 0
    }
}

# Mostrar el state
print("=== AGENT STATE (Patrón LangGraph) ===")
print(f"User ID: {agent_state['user_id']}")
print(f"Current step: {agent_state['current_step']}")
print(f"Messages count: {len(agent_state['messages'])}")

print("\n=== MENSAJES ===")
for i, msg in enumerate(agent_state["messages"]):
    print(f"{i+1}. [{msg['role']}]: {msg['content']}")
````

### 🔗 Conexión con LangGraph

Lo que acabas de ver **ES** el patrón real que usa LangGraph. Cuando definas un agente, usarás `TypedDict` para dar estructura formal a este state:

```python
# Así se define un State en LangGraph (lo verás pronto)
from typing import TypedDict, Annotated
from langgraph.graph.message import add_messages

class AgentState(TypedDict):
    messages: Annotated[list, add_messages]
    user_id: str
    current_step: str
    metadata: dict
```

Por ahora, domina las listas y diccionarios. El `TypedDict` simplemente agrega verificación de tipos.

# PARTE 5: CONDICIONALES (if/elif/else)

## ¿Qué es un condicional?
Un **condicional** permite ejecutar código diferente dependiendo de si una condición es verdadera o falsa.

## ¿Por qué los agentes necesitan condicionales?
- **Decidir qué responder** según el tipo de mensaje
- **Verificar permisos** antes de ejecutar acciones
- **Validar datos** antes de procesarlos

## Sintaxis
```python
if condicion:
    # código si condición es True
elif otra_condicion:
    # código si otra_condicion es True
else:
    # código si ninguna condición es True
```

## Operadores de comparación
| Operador | Significado | Ejemplo |
|----------|-------------|---------|
| `==` | Igual a | `x == 5` |
| `!=` | Diferente de | `x != 5` |
| `<` | Menor que | `x < 5` |
| `>` | Mayor que | `x > 5` |
| `<=` | Menor o igual | `x <= 5` |
| `>=` | Mayor o igual | `x >= 5` |
| `in` | Está contenido en | `"a" in "abc"` |
| `not in` | No está contenido en | `"x" not in "abc"` |

## Operadores lógicos
| Operador | Significado | Ejemplo |
|----------|-------------|---------|
| `and` | Y (ambas deben ser True) | `x > 0 and x < 10` |
| `or` | O (al menos una True) | `x < 0 or x > 10` |
| `not` | Negación | `not is_active` |

````python
# CONDICIONALES EN MEMORIA DE AGENTES

# Estado actual de la memoria
memory_size = 85
MAX_MEMORY = 100
is_memory_enabled = True

# DECISIÓN: ¿La memoria está disponible?
if not is_memory_enabled:
    print("❌ Memoria deshabilitada. No se pueden guardar mensajes.")
elif memory_size >= MAX_MEMORY:
    print("⚠️ Memoria llena. Se eliminarán mensajes antiguos.")
elif memory_size >= MAX_MEMORY * 0.8:  # 80% de capacidad
    print(f"⚠️ Memoria casi llena: {memory_size}/{MAX_MEMORY}")
else:
    print(f"✅ Memoria disponible: {memory_size}/{MAX_MEMORY}")

print("\n" + "="*50 + "\n")

# DECISIÓN: ¿Qué tipo de mensaje es?
message = "Usuario: ¿Cuál es el clima?"

if message.startswith("Usuario:"):
    role = "user"
    content = message.replace("Usuario:", "").strip()
    print(f"Mensaje de USUARIO detectado: '{content}'")
elif message.startswith("Agente:"):
    role = "assistant"
    content = message.replace("Agente:", "").strip()
    print(f"Mensaje de AGENTE detectado: '{content}'")
else:
    role = "unknown"
    content = message
    print(f"Mensaje de origen DESCONOCIDO: '{content}'")

print("\n" + "="*50 + "\n")

# DECISIÓN: ¿El mensaje requiere búsqueda en memoria?
keywords_requiring_memory = ["recuerdas", "dijiste", "anterior", "antes"]

needs_memory_search = False
for keyword in keywords_requiring_memory:
    if keyword in content.lower():
        needs_memory_search = True
        break  # Ya encontramos uno, no necesitamos seguir buscando

if needs_memory_search:
    print("🔍 Este mensaje requiere buscar en la memoria del agente")
else:
    print("✅ Este mensaje no requiere búsqueda en memoria")
````

# PARTE 6: LOOPS (for, while)

## ¿Qué es un loop?
Un **loop** (bucle) permite ejecutar código repetidamente.

## ¿Por qué los agentes necesitan loops?
- **Iterar sobre memoria**: Recorrer historial de mensajes
- **Buscar información**: Encontrar mensajes específicos
- **Procesar listas**: Aplicar operaciones a múltiples elementos

## Tipos de loops

### For loop - Iterar sobre una secuencia
```python
for elemento in secuencia:
    # código a ejecutar para cada elemento
```

### While loop - Repetir mientras condición sea True
```python
while condicion:
    # código a ejecutar mientras condición es True
```

## Funciones útiles para loops
| Función | Descripción | Ejemplo |
|---------|-------------|---------|
| `range(n)` | Genera números 0 a n-1 | `range(5)` → 0,1,2,3,4 |
| `range(a, b)` | Genera números a hasta b-1 | `range(2, 5)` → 2,3,4 |
| `enumerate(lista)` | Índice y elemento | `for i, x in enumerate(lista)` |
| `zip(a, b)` | Combina dos listas | `for x, y in zip(a, b)` |

## Control de loops
| Keyword | Efecto |
|---------|--------|
| `break` | Termina el loop inmediatamente |
| `continue` | Salta a la siguiente iteración |

````python
# LOOPS PARA BUSCAR EN MEMORIA

# Memoria del agente con mensajes anteriores
memory = [
    {"role": "user", "content": "Hola, soy Carlos"},
    {"role": "assistant", "content": "¡Hola Carlos! ¿Cómo puedo ayudarte?"},
    {"role": "user", "content": "Me gusta el fútbol"},
    {"role": "assistant", "content": "Interesante, ¿cuál es tu equipo favorito?"},
    {"role": "user", "content": "Real Madrid"},
    {"role": "assistant", "content": "¡Gran equipo! Recordaré que eres fan del Real Madrid."},
    {"role": "user", "content": "¿Recuerdas mi nombre?"}
]

# FOR LOOP: Mostrar toda la memoria con índice
print("=== MEMORIA COMPLETA ===")
for index, mensaje in enumerate(memory):
    print(f"{index}: [{mensaje['role']}] {mensaje['content']}")

# FOR LOOP: Buscar información específica (nombre del usuario)
print("\n=== BUSCANDO NOMBRE DEL USUARIO ===")
user_name = None
for mensaje in memory:
    if mensaje["role"] == "user" and "soy" in mensaje["content"].lower():
        # Extraer nombre (simplificado)
        words = mensaje["content"].split()
        for i, word in enumerate(words):
            if word.lower() == "soy" and i + 1 < len(words):
                user_name = words[i + 1]
                break
    if user_name:
        break  # Ya encontramos el nombre, salir del loop

if user_name:
    print(f"✅ Nombre encontrado: {user_name}")
else:
    print("❌ No se encontró el nombre")

# FOR LOOP: Contar mensajes por rol
print("\n=== ESTADÍSTICAS DE MEMORIA ===")
user_messages = 0
assistant_messages = 0

for mensaje in memory:
    if mensaje["role"] == "user":
        user_messages += 1
    elif mensaje["role"] == "assistant":
        assistant_messages += 1

print(f"Mensajes de usuario: {user_messages}")
print(f"Mensajes de asistente: {assistant_messages}")
print(f"Total: {len(memory)}")
````

````python
# WHILE LOOP: Limitar memoria cuando está llena

# Simular memoria grande
large_memory = [f"Mensaje {i}" for i in range(15)]
MAX_SIZE = 10

print(f"=== GESTIÓN DE MEMORIA CON WHILE ===")
print(f"Tamaño inicial: {len(large_memory)}")
print(f"Límite máximo: {MAX_SIZE}")

# Eliminar mensajes antiguos hasta que esté dentro del límite
removed_count = 0
while len(large_memory) > MAX_SIZE:
    removed_message = large_memory.pop(0)  # Eliminar el más antiguo
    removed_count += 1
    print(f"  Eliminado: '{removed_message}'")

print(f"\nMensajes eliminados: {removed_count}")
print(f"Tamaño final: {len(large_memory)}")
print(f"Memoria actual: {large_memory}")

# LIST COMPREHENSION (forma compacta de crear listas con for)
print("\n=== LIST COMPREHENSION ===")

# Forma tradicional
mensajes_usuario_tradicional = []
for msg in memory:
    if msg["role"] == "user":
        mensajes_usuario_tradicional.append(msg["content"])

# Forma con list comprehension (más pythonic)
mensajes_usuario_comprehension = [msg["content"] for msg in memory if msg["role"] == "user"]

print("Mensajes de usuario (list comprehension):")
for msg in mensajes_usuario_comprehension:
    print(f"  • {msg}")
````

# PARTE 7: FUNCIONES (def)

## ¿Qué es una función?
Una **función** es un bloque de código reutilizable que realiza una tarea específica.

## ¿Por qué los agentes necesitan funciones?
- **Modularidad**: Separar operaciones de memoria en piezas manejables
- **Reutilización**: No repetir código
- **Testing**: Probar cada operación individualmente
- **Mantenimiento**: Cambiar comportamiento en un solo lugar

## Sintaxis básica
```python
def nombre_funcion(parametro1, parametro2):
    """Docstring: Descripción de qué hace la función"""
    # código
    return resultado
```

## Anatomía de una función
```python
def calcular_promedio(numeros):  # Nombre y parámetros
    """Calcula el promedio de una lista de números."""  # Docstring
    if not numeros:  # Validación
        return 0
    suma = sum(numeros)  # Lógica
    promedio = suma / len(numeros)
    return promedio  # Valor de retorno
```

````python
# FUNCIONES PARA OPERACIONES DE MEMORIA

def create_message(role, content, user_id=None):
    """
    Crea un mensaje estructurado para la memoria del agente.
    
    Args:
        role: "user" o "assistant"
        content: El texto del mensaje
        user_id: Identificador opcional del usuario
    
    Returns:
        dict: Mensaje estructurado con timestamp
    """
    from datetime import datetime
    
    message = {
        "role": role,
        "content": content,
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }
    
    if user_id:
        message["user_id"] = user_id
    
    return message

# Usar la función
msg1 = create_message("user", "Hola, ¿cómo estás?", "user_123")
msg2 = create_message("assistant", "¡Hola! Estoy bien, gracias.")

print("=== MENSAJES CREADOS ===")
print(f"Mensaje 1: {msg1}")
print(f"Mensaje 2: {msg2}")
````

````python
# MÁS FUNCIONES DE MEMORIA

def add_to_memory(memory, message, max_size=100):
    """
    Agrega un mensaje a la memoria, eliminando antiguos si es necesario.
    
    Args:
        memory: Lista de mensajes (se modifica in-place)
        message: Mensaje a agregar
        max_size: Tamaño máximo de la memoria
    
    Returns:
        int: Número de mensajes eliminados
    """
    removed = 0
    
    # Eliminar mensajes antiguos si la memoria está llena
    while len(memory) >= max_size:
        memory.pop(0)
        removed += 1
    
    # Agregar nuevo mensaje
    memory.append(message)
    
    return removed


def search_in_memory(memory, keyword, role=None):
    """
    Busca mensajes que contengan una palabra clave.
    
    Args:
        memory: Lista de mensajes
        keyword: Palabra a buscar
        role: Opcional - filtrar por rol ("user" o "assistant")
    
    Returns:
        list: Mensajes que coinciden con la búsqueda
    """
    results = []
    
    for message in memory:
        # Filtrar por rol si se especifica
        if role and message.get("role") != role:
            continue
        
        # Buscar keyword en contenido
        if keyword.lower() in message.get("content", "").lower():
            results.append(message)
    
    return results


def get_recent_context(memory, limit=5):
    """
    Obtiene los últimos N mensajes como contexto.
    
    Args:
        memory: Lista de mensajes
        limit: Número de mensajes a obtener
    
    Returns:
        list: Últimos mensajes
    """
    return memory[-limit:] if len(memory) > limit else memory.copy()


# PROBAR LAS FUNCIONES
print("=== PRUEBA DE FUNCIONES DE MEMORIA ===\n")

# Crear memoria y agregar mensajes
agent_memory = []

messages_to_add = [
    ("user", "Hola, soy Ana"),
    ("assistant", "¡Hola Ana! ¿Cómo puedo ayudarte?"),
    ("user", "Me encanta la música jazz"),
    ("assistant", "¡Genial! El jazz es un género muy rico. ¿Tienes algún artista favorito?"),
    ("user", "Sí, me gusta mucho Miles Davis"),
    ("assistant", "Miles Davis es legendario. Recordaré que te gusta el jazz y Miles Davis."),
    ("user", "¿Qué recuerdas de mí?")
]

for role, content in messages_to_add:
    msg = create_message(role, content, "user_456")
    removed = add_to_memory(agent_memory, msg, max_size=10)
    if removed > 0:
        print(f"⚠️ Se eliminaron {removed} mensajes antiguos")

print(f"Mensajes en memoria: {len(agent_memory)}")

# Buscar información
print("\n--- Buscando 'música' ---")
results = search_in_memory(agent_memory, "música")
for r in results:
    print(f"  [{r['role']}]: {r['content']}")

print("\n--- Buscando mensajes del usuario sobre 'Miles' ---")
results = search_in_memory(agent_memory, "Miles", role="user")
for r in results:
    print(f"  [{r['role']}]: {r['content']}")

# Obtener contexto reciente
print("\n--- Últimos 3 mensajes (contexto) ---")
context = get_recent_context(agent_memory, limit=3)
for msg in context:
    print(f"  [{msg['role']}]: {msg['content']}")
````

# PARTE 8: JSON - Persistir Memoria en Archivos

## ¿Qué es JSON?
**JSON** (JavaScript Object Notation) es un formato de texto para almacenar y transmitir datos estructurados. Es legible por humanos y máquinas.

## ¿Por qué los agentes necesitan JSON?
- **Persistencia**: Guardar memoria cuando el agente se cierra
- **Recuperación**: Cargar memoria cuando el agente inicia
- **Configuración**: Almacenar configuraciones en archivos
- **APIs**: La mayoría de APIs usan JSON para comunicarse

## Estructura JSON
JSON soporta los mismos tipos que Python:
```json
{
    "string": "texto",
    "number": 42,
    "float": 3.14,
    "boolean": true,
    "null": null,
    "array": [1, 2, 3],
    "object": {"key": "value"}
}
```

## Librería `json` (built-in)
| Función | Descripción |
|---------|-------------|
| `json.dumps(obj)` | Convierte objeto Python → string JSON |
| `json.loads(str)` | Convierte string JSON → objeto Python |
| `json.dump(obj, file)` | Escribe objeto Python → archivo JSON |
| `json.load(file)` | Lee archivo JSON → objeto Python |

**Nota**: Esta librería viene incluida con Python, no necesitas instalar nada.

````python
# JSON: PERSISTIR Y RECUPERAR MEMORIA

import json

# Nuestra memoria actual (de ejemplos anteriores)
memory_to_save = [
    {"role": "user", "content": "Hola, soy Ana", "timestamp": "2024-01-15 10:00:00"},
    {"role": "assistant", "content": "¡Hola Ana!", "timestamp": "2024-01-15 10:00:02"},
    {"role": "user", "content": "Me gusta el jazz", "timestamp": "2024-01-15 10:01:00"},
]

# 1. CONVERTIR A STRING JSON (para ver/transmitir)
json_string = json.dumps(memory_to_save)
print("=== MEMORIA COMO STRING JSON ===")
print(json_string)

# 2. CONVERTIR CON FORMATO LEGIBLE
json_pretty = json.dumps(memory_to_save, indent=2, ensure_ascii=False)
print("\n=== MEMORIA FORMATEADA ===")
print(json_pretty)

# 3. GUARDAR EN ARCHIVO
filename = "agent_memory.json"
with open(filename, "w", encoding="utf-8") as file:
    json.dump(memory_to_save, file, indent=2, ensure_ascii=False)
print(f"\n✅ Memoria guardada en '{filename}'")

# 4. CARGAR DESDE ARCHIVO
with open(filename, "r", encoding="utf-8") as file:
    loaded_memory = json.load(file)
print(f"\n=== MEMORIA CARGADA DESDE ARCHIVO ===")
print(f"Mensajes recuperados: {len(loaded_memory)}")
for msg in loaded_memory:
    print(f"  [{msg['role']}]: {msg['content']}")
````

````python
# FUNCIONES COMPLETAS DE PERSISTENCIA

import json
import os

def save_memory(memory, filename="agent_memory.json"):
    """
    Guarda la memoria del agente en un archivo JSON.
    
    Args:
        memory: Lista de mensajes
        filename: Nombre del archivo
    
    Returns:
        bool: True si se guardó correctamente
    """
    try:
        with open(filename, "w", encoding="utf-8") as file:
            json.dump(memory, file, indent=2, ensure_ascii=False)
        return True
    except Exception as e:
        print(f"Error guardando memoria: {e}")
        return False


def load_memory(filename="agent_memory.json"):
    """
    Carga la memoria del agente desde un archivo JSON.
    
    Args:
        filename: Nombre del archivo
    
    Returns:
        list: Memoria cargada, o lista vacía si no existe
    """
    # Verificar si el archivo existe
    if not os.path.exists(filename):
        print(f"Archivo '{filename}' no encontrado. Iniciando memoria vacía.")
        return []
    
    try:
        with open(filename, "r", encoding="utf-8") as file:
            memory = json.load(file)
        print(f"✅ Memoria cargada: {len(memory)} mensajes")
        return memory
    except Exception as e:
        print(f"Error cargando memoria: {e}")
        return []


# SIMULAR CICLO DE VIDA DEL AGENTE

print("=== SIMULACIÓN: AGENTE INICIA ===")
# 1. Cargar memoria existente
agent_memory = load_memory("test_memory.json")

print("\n=== SIMULACIÓN: NUEVA CONVERSACIÓN ===")
# 2. Agregar nuevos mensajes
new_messages = [
    create_message("user", "¿Qué hora es?", "user_789"),
    create_message("assistant", "Son las 3:00 PM"),
]
for msg in new_messages:
    add_to_memory(agent_memory, msg)
    print(f"  + Agregado: [{msg['role']}] {msg['content']}")

print("\n=== SIMULACIÓN: AGENTE SE CIERRA ===")
# 3. Guardar memoria antes de cerrar
success = save_memory(agent_memory, "test_memory.json")
if success:
    print("✅ Memoria persistida correctamente")

print("\n=== SIMULACIÓN: AGENTE REINICIA ===")
# 4. Cargar memoria (como si fuera un nuevo inicio)
recovered_memory = load_memory("test_memory.json")
print("\nMemoria recuperada:")
for msg in recovered_memory:
    print(f"  [{msg['role']}]: {msg['content']}")
````

````python
# SISTEMA COMPLETO: AgentMemory
# Este es código que realmente usarías en un proyecto de agentes

import json
import os
from datetime import datetime


# ============================================
# CONFIGURACIÓN (Variables)
# ============================================
MEMORY_FILE = "agent_memory.json"
MAX_MEMORY_SIZE = 100
DEFAULT_CONTEXT_SIZE = 5


# ============================================
# FUNCIONES DE MEMORIA
# ============================================

def create_message(role, content, user_id=None):
    """Crea un mensaje estructurado para la memoria."""
    message = {
        "role": role,
        "content": content,
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }
    if user_id:
        message["user_id"] = user_id
    return message


def add_to_memory(memory, role, content, user_id=None, max_size=MAX_MEMORY_SIZE):
    """Agrega un mensaje a la memoria, gestionando el límite de tamaño."""
    # Crear mensaje
    message = create_message(role, content, user_id)
    
    # Eliminar antiguos si es necesario
    while len(memory) >= max_size:
        memory.pop(0)
    
    # Agregar nuevo mensaje
    memory.append(message)
    return message


def search_memory(memory, keyword, role=None, limit=None):
    """Busca mensajes que contengan una palabra clave."""
    results = []
    
    for message in memory:
        # Filtrar por rol si se especifica
        if role and message.get("role") != role:
            continue
        
        # Buscar keyword
        if keyword.lower() in message.get("content", "").lower():
            results.append(message)
            
            # Limitar resultados si se especifica
            if limit and len(results) >= limit:
                break
    
    return results


def get_context(memory, limit=DEFAULT_CONTEXT_SIZE):
    """Obtiene los últimos N mensajes como contexto."""
    return memory[-limit:] if len(memory) > limit else memory.copy()


def get_user_messages(memory, user_id):
    """Obtiene todos los mensajes de un usuario específico."""
    return [msg for msg in memory if msg.get("user_id") == user_id]


def save_memory(memory, filename=MEMORY_FILE):
    """Guarda la memoria en un archivo JSON."""
    try:
        with open(filename, "w", encoding="utf-8") as f:
            json.dump(memory, f, indent=2, ensure_ascii=False)
        return True
    except Exception as e:
        print(f"Error al guardar: {e}")
        return False


def load_memory(filename=MEMORY_FILE):
    """Carga la memoria desde un archivo JSON."""
    if not os.path.exists(filename):
        return []
    
    try:
        with open(filename, "r", encoding="utf-8") as f:
            return json.load(f)
    except Exception as e:
        print(f"Error al cargar: {e}")
        return []


def get_memory_stats(memory):
    """Obtiene estadísticas de la memoria."""
    user_count = sum(1 for msg in memory if msg.get("role") == "user")
    assistant_count = sum(1 for msg in memory if msg.get("role") == "assistant")
    
    return {
        "total_messages": len(memory),
        "user_messages": user_count,
        "assistant_messages": assistant_count,
        "capacity_used": f"{len(memory)}/{MAX_MEMORY_SIZE}",
        "is_full": len(memory) >= MAX_MEMORY_SIZE
    }


# ============================================
# DEMOSTRACIÓN DEL SISTEMA COMPLETO
# ============================================

print("=" * 60)
print("SISTEMA DE MEMORIA DE AGENTE - DEMOSTRACIÓN")
print("=" * 60)

# 1. Inicializar memoria
memory = load_memory("demo_memory.json")
print(f"\n1. Memoria inicializada: {len(memory)} mensajes previos")

# 2. Simular conversación
print("\n2. Agregando conversación...")
conversations = [
    ("user", "Hola, me llamo Pedro", "user_001"),
    ("assistant", "¡Hola Pedro! ¿En qué puedo ayudarte hoy?", None),
    ("user", "Quiero aprender sobre inteligencia artificial", "user_001"),
    ("assistant", "¡Excelente elección! La IA es un campo fascinante. ¿Por dónde te gustaría empezar?", None),
    ("user", "¿Qué es machine learning?", "user_001"),
    ("assistant", "Machine Learning es una rama de la IA donde los sistemas aprenden de datos.", None),
]

for role, content, user_id in conversations:
    msg = add_to_memory(memory, role, content, user_id)
    print(f"   + [{role}]: {content[:50]}...")

# 3. Estadísticas
print("\n3. Estadísticas de memoria:")
stats = get_memory_stats(memory)
for key, value in stats.items():
    print(f"   - {key}: {value}")

# 4. Búsqueda
print("\n4. Búsqueda en memoria:")
print("   Buscando 'inteligencia':")
results = search_memory(memory, "inteligencia")
for r in results:
    print(f"   → [{r['role']}]: {r['content'][:50]}...")

# 5. Contexto para el agente
print("\n5. Contexto reciente (últimos 3 mensajes):")
context = get_context(memory, limit=3)
for msg in context:
    print(f"   [{msg['role']}]: {msg['content'][:50]}...")

# 6. Persistencia
print("\n6. Guardando memoria...")
save_memory(memory, "demo_memory.json")
print("   ✅ Memoria guardada en 'demo_memory.json'")

print("\n" + "=" * 60)
print("FIN DE LA DEMOSTRACIÓN")
print("=" * 60)
````
